Sum of weights

```python
asia = df[df['origin'] == 'Asia']

X = asia[['vehicle_weight', 'model_year']].iloc[:7].to_numpy()
```

```python
XTX = X.T @ X
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv @ X.T @ y
```

```python
w.sum()
```
